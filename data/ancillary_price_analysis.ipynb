```python
import pandas as pd
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv("NEW_Ercot_mcpc_comparison.csv")

df.head()
```

```python
df["time"] = pd.to_datetime(df["interval_start_local"])
df["month_day"] = df["time"].dt.strftime("%m-%d")


fig1 = (
    df.query("as_type == 'NSPIN' ") # filter to just NSPIN
    .assign(year = lambda x: df["time"].dt.year) # add a year column
    .assign(date = lambda x: df["month_day"])  #date col (month/ day)
    .groupby(["date", "year"])
    .agg(med_daily = ("mcpc", lambda x: x.median()))
)
df["year"] = df["time"].dt.year

sns.lineplot( 
     data = fig1,
     x = "date",
     y = "med_daily",
     hue = "year"
 )

fig1.head()
```
